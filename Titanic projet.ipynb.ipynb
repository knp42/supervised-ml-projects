### Handle missing values

```python
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, confusion_matrix
```

```python
df.dtypes
```

```python
numeric_col = ['survived','pclass','age','sibsp','parch','fare']
category_col = ['sex','embarked','class','who','adult_male','deck','embark_town','alive','alone']

# fit a missing value
numeric_impute = SimpleImputer(strategy = 'mean')
df[numeric_col]= numeric_impute.fit_transform(df[numeric_col])

category_impute = SimpleImputer(strategy = 'most_frequent')
df[category_col] = category_impute.fit_transform(df[category_col])

# encoder the numeric and category values
df_encoded = pd.get_dummies(df, columns = category_col , drop_first = True)
df_encoded
```

```python
df_encoded.dtypes
```

```python
cols_to_drop = ['alive_yes', 'class_Second', 'class_Third', 
                 'who_man', 'who_woman', 'adult_male_True',
                 'embark_town_Queenstown', 'embark_town_Southampton',
                 'deck_B', 'deck_C', 'deck_D', 'deck_E', 'deck_F', 'deck_G']

df_encoded = df_encoded.drop(columns=cols_to_drop)
print(df_encoded.columns.tolist())
```

```python
X = df_encoded.drop('survived', axis=1)
y = df_encoded['survived']
```

```python
y.value_counts()
```

```python
X.isnull().sum()
```

```python
X = df_encoded.drop('survived', axis=1)
y = df_encoded['survived']


# Normalization
scaler = StandardScaler()
X_scaled =scaler.fit_transform(X)

# select the model
X_train,X_test,y_train,y_test = train_test_split(X_scaled,y, test_size=0.2, random_state = 42 )

# instantiate the model
models ={
    'Logistic_Reg': LogisticRegression(max_iter=1000),
    'random_forest': RandomForestClassifier(n_estimators=100, random_state=42),
    'svc': SVC(kernel='linear',probability=True,random_state=42)
}

# fitting each model
result = {}
for name, model in models.items():
    model.fit(X_train,y_train)
    pred = model.predict(X_test)
    prob = model.predict_proba(X_test)[:,1] if hasattr(model,'predict_proba') else None

    result[name] = {
        'accuracy score': accuracy_score(y_test,pred),
        'F1-score': f1_score(y_test,pred),
        'ROC-AUC': roc_auc_score(y_test,prob) if prob is not None else "N/A",
        'Confusion Matrix': confusion_matrix(y_test,pred)
    }
for name,metrics in result.items():
    print(f"\n=== {name} ===")
    for  metric,value in metrics.items():
        print(f"{metric}: {value}")
```

```python
from sklearn.model_selection import cross_validate
```

```python
scoring = ['accuracy', 'f1','roc_auc']
cv_result = {}
for name, model in models.items():
    scores = cross_validate(model, X_scaled,y, cv=5, scoring=scoring)
    cv_result[name]={
        metric: scores[f'test_{metric}'].mean() for metric in scoring
    }
for name, res in cv_result.items():
    print(f"\n==={name}===")
    for metric, value in res.items():
        print(f"{metric}: {value:.4f}")
```

### ROC curve comparaison of titanic prediction models

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc
```

```python
plt.figure(figsize=(8,6))
for name,model in models.items():
    model.fit(X_train,y_train)
    y_prob = model.predict_proba(X_test)[:,1]

    fpr,tpr,tresholds = roc_curve(y_test,y_prob)
    roc_auc = auc(fpr,tpr)
    plt.plot(fpr,tpr, label=f"{name}(AUC ={roc_auc:.3f})")

plt.plot([0,1],[0,1], linestyle = '--', color='gray', label='Random guess')
plt.xlabel('False positive rate')
plt.ylabel('True positive rate')
plt.title('Roc_curve comparaison Titanic prediction models')
plt.legend(loc= 'lower right')
plt.grid(alpha = 0.3)
plt.show()
```
